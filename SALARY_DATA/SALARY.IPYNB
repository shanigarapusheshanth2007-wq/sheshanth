```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("SALARY_DATA.csv")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import (mean_absolute_error,mean_squared_error,r2_score)
df
X=df.loc[:,["YearsExperience"]]
y=df.loc[:,["Salary"]]
x_train,x_test,y_train,y_test=train_test_split(X,y,random_state=3)
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
a=np.arange(0,10,0.2)
plt.scatter(x=df.loc[:,["YearsExperience"]],y=df.loc[:,["Salary"]],label="actual values")
plt.scatter(x=a,y=np.round(model.predict(a.reshape(-1,1),),2),label="predicted") 
plt.ylabel("output")
plt.xlabel("experience")
plt.legend()
```

Now using the polynomail features

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=5)
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import (mean_absolute_error,mean_squared_error,r2_score)
df
X=df.loc[:,["YearsExperience"]]
y=df.loc[:,["Salary"]]
x_train,x_test,y_train,y_test=train_test_split(X,y,random_state=3)
x_train=poly.fit_transform(x_train)
x_test=poly.transform(x_test)
model1=LinearRegression()
model1.fit(x_train,y_train)
y_pred=model1.predict(x_test)
```

```python
a=np.arange(0,12,0.01)
a_poly = poly.transform(a.reshape(-1, 1))
plt.scatter(x=df.loc[:,["YearsExperience"]],y=df.loc[:,["Salary"]],label="actual values")
plt.scatter(x=a,y=np.round(model1.predict(a_poly,),2),label="predicted") 
plt.ylabel("output")
plt.xlabel("experience")
plt.legend()
```
